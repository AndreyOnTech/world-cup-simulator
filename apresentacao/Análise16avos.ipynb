```python
import json
from pathlib import Path
```

```python
resultado_file = Path("./../sistema_elo/resultado.json")

with open(resultado_file, "r", encoding="utf-8") as f:
    data = json.load(f)

groups = {
    "A": ["Mexico", "South Africa", "Korea Republic", "Czechia"],
    "B": ["Canada", "Bosnia and Herzegovina", "Qatar", "Switzerland"],
    "C": ["Brazil", "Morocco", "Haiti", "Scotland"],
    "D": ["USA", "Paraguay", "Australia", "Türkiye"],
    "E": ["Germany", "Curaçao", "Côte d'Ivoire", "Ecuador"],
    "F": ["Netherlands", "Japan", "Sweden", "Tunisia"],
    "G": ["Belgium", "Egypt", "IR Iran", "New Zealand"],
    "H": ["Spain", "Cabo Verde", "Saudi Arabia", "Uruguay"],
    "I": ["France", "Senegal", "Iraq", "Norway"],
    "J": ["Argentina", "Algeria", "Austria", "Jordan"],
    "K": ["Portugal", "Congo DR", "Uzbekistan", "Colombia"],
    "L": ["England", "Croatia", "Ghana", "Panama"],
}

terceiros = []
ranking_por_grupo = {}

for grupo, teams in groups.items():

    ranking = sorted(
        teams,
        key=lambda team: data[team]["classificado"],
        reverse=True
    )

    ranking_por_grupo[grupo] = ranking

    terceiros.append((
        grupo,
        ranking[2],
        data[ranking[2]]["classificado"]
    ))

melhores_terceiros = {
    team
    for _, team, _ in sorted(
        terceiros,
        key=lambda x: x[2],
        reverse=True
    )[:8]
}

classificados_simulacao = {}

for grupo, ranking in ranking_por_grupo.items():

    classificados_simulacao[grupo] = []

    # 1º e 2º colocados
    classificados_simulacao[grupo].append(ranking[0])
    classificados_simulacao[grupo].append(ranking[1])

    # Melhor terceiro, se houver
    if ranking[2] in melhores_terceiros:
        classificados_simulacao[grupo].append(ranking[2])

for grupo, times in classificados_simulacao.items():
    print(f"Grupo {grupo}")
    for i, time in enumerate(times, start=1):
        print(f"  {i}º {time}")
    print()
```

```python
resultado_file = Path("./../16avos_classificados.json")
with open(resultado_file, "r", encoding="utf-8") as f:
    data = json.load(f)

total_possivel_classificar = 32
pontuacao_acerto = 0

for group, teams in data['groups'].items():
    for team in teams:
        if (team['team'] in classificados_simulacao[group]):
            pontuacao_acerto = pontuacao_acerto+1

print(pontuacao_acerto)

percentual_acerto = (pontuacao_acerto*100)/total_possivel_classificar
print(f"{percentual_acerto:.2f}%")
```

```python
resultado_file = Path("./../16avos_classificados.json")
with open(resultado_file, "r", encoding="utf-8") as f:
    data = json.load(f)

total_possivel_classificar = 64
pontuacao_acerto = 0

for grupo, teams in data["groups"].items():

    simulacao = classificados_simulacao[grupo]

    for posicao, team in enumerate(teams):

        nome = team["team"]

        if nome in simulacao:
            pontuacao_acerto += 1

            if posicao < len(simulacao) and simulacao[posicao] == nome:
                pontuacao_acerto += 1

print(f"Pontuação: {pontuacao_acerto}/{total_possivel_classificar}")

percentual_acerto = (pontuacao_acerto*100)/total_possivel_classificar
print(f"Aproveitamento: {percentual_acerto:.2f}%")
```
